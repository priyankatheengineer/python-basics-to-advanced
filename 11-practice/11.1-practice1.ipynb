```python
## Find top 3 most frequent elements in a list
nums = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5]

# Step 1: Count frequency
freq = {x: nums.count(x) for x in set(nums)}
print("Frequencies:", freq)

# Step 2: Sort by frequency (descending) and get top 3 keys
sorted_keys = list(dict(sorted(freq.items(), key=lambda x: x[1], reverse=True)).keys())
top_3 = sorted_keys[:3]

print("Top 3 most frequent elements:", top_3)
```

### This is called a Linear Congruential Generator (LCG) — a simple algorithm to generate pseudo-random numbers.
seed = (seed * 9301 + 49297) % 233280
Xn+1 = (aXn + c) mod m
Where:

a = 9301

c = 49297

m = 233280

```python
## Generate random numbers without random module

import time
def pseudo_random(seed=1):
    seed = (seed*9301+49297) % 233280
    return seed/233280.0
print(pseudo_random(int(time.time())))
```

```python
## 𝐆𝐢𝐯𝐞𝐧 𝐚 𝐬𝐭𝐫𝐢𝐧𝐠 𝐩𝐡𝐫𝐚𝐬𝐞, 𝐫𝐞𝐭𝐮𝐫𝐧 𝐓𝐫𝐮𝐞 𝐢𝐟 𝐢𝐭 𝐢𝐬 𝐚 𝐩𝐚𝐥𝐢𝐧𝐝𝐫𝐨𝐦𝐞, 𝐨𝐭𝐡𝐞𝐫𝐰𝐢𝐬𝐞 𝐫𝐞𝐭𝐮𝐫𝐧 𝐅𝐚𝐥𝐬𝐞. 

def is_palindrome(sentence):
    # Keep only letters and numbers, make lowercase
    cleaned = ''.join(ch.lower() for ch in sentence if ch.isalnum())
    return cleaned == cleaned[::-1]

# Example usage
sentence1 = "A man, a plan, a canal, Panama!"
sentence2 = "Hello World"

print(is_palindrome(sentence1))  # True
print(is_palindrome(sentence2))  # False
```

```python
## palindrome checker that keeps all characters, including letters, numbers, and symbols
def is_palindrome(sentence):
    # Compare the sentence with its reverse
    return sentence == sentence[::-1]

# Examples
sentence1 = "!23$32!"
sentence2 = "A man, a plan, a canal, Panama!"
sentence3 = "Hello World"

print(is_palindrome(sentence1))  # True
print(is_palindrome(sentence2))  # False (because spaces and capitalization matter)
print(is_palindrome(sentence3))  # False
```

A context manager is a Python construct that automatically manages resources.

It sets up a resource, runs a block of code, and cleans up the resource automatically.

Commonly used for things like opening files, acquiring locks, or connecting to databases.

Key idea: “Do setup, run your code, then cleanup automatically.”

```python
## Custome context manager

class MyContext:
    def __enter__(self):
        print("Entering the Context")
        return "Resource"
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting the Context")
        if exc_type:
            print(f"Exception: {exc_value}")
        return True

with MyContext() as res:
    print(f"Using {res}")
```

A regular expression (regex) is a special sequence of characters that defines a search pattern.

Used to match, search, and manipulate strings.

Can be simple (find a word) or complex (validate emails, extract phone numbers).

Example:

Pattern: r"\d+" → matches one or more digits in a string.

```python
## mtach() - Checks if the pattern matches at the beginning of the string.
## Returns a match object if found, else None

import re
sentence = "Python is programming language"
pattern = r"Python"

result = re.match(pattern,sentence)
print(result)
print(result.group())
print(result.start())
print(result.end())
```

```python
## search() - Searches anywhere in the string. Returns the first match as a match object.

import re
sentence = "I love Python"
pattern = r"Python"

result = re.search(pattern, sentence)
print(result)
print(result.group())
```

```python
## findall() - Returns all matches in the string as a list.

import re

sentence = "Python3, Java 8, C++ 14"
numbers = re.findall(r"\d+",sentence)
print(numbers)
```

```python
## finditer() - Returns an iterator of match objects for all matches. Useful if you need positions of matches.

sentence = "Python3, Java 8, C++ 14"

for i in re.finditer(r"\d+", sentence):
    print(i.group(), "at", i.start())
```

```python
## split() - Splits a string by pattern instead of a fixed delimiter.

sentence = "Python Javascript DBMS"
result = re.split(r' ', sentence)
print(result)
```

```python
## sub() - Replaces pattern matches with a replacement string.
sentence = "I have 4 cats and 6 dogs"

result = re.sub(r"\d+", "$", sentence)
print(result)
```

```python
## fullmatch() - Checks if the entire string matches the pattern.

sentence = "Python3.8 is good"
pattern = r"[\w.]+\s\w+\s\w+"
result = re.fullmatch(pattern, sentence)
print(result)
print(result.group())
```

```python
## Validate email

email = "hello.123@gmail.com"
pattern = r"[\w.]+@[a-zA-Z]+\.[a-zA-z]+"
result = re.findall(pattern, email)
print(result)
```

```python
text = "John: 25"
pattern = r"(\w+): (\d+)"
m = re.match(pattern, text)
print(m.group())     # 'John: 25' → whole match
print(m.group(1))    # 'John' → first group (\w+)
print(m.group(2))    # '25' → second group (\d+)
```

yield is used in functions to make them generators.

A generator is an iterator that produces values one at a time, instead of computing them all at once and storing in memory.

Using yield pauses the function, remembers its state, and resumes from there the next time it’s called.

```python
def simple_gen():
    yield 1
    yield 2
    yield 3

gen = simple_gen()
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
# next(gen) → StopIteration
```

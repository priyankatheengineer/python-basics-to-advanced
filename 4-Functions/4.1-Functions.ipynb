```python
### Variable length arguments

## Positional and keyword arguments
def print_number(*args):
    for n in args:
        print(n)

print_number(1,2,3,4,5,6,7,8)
```

```python
### Keyword Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Krish",age="32",country="India")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(1,2,3,4,5,name="Krish",age="32",country="India")
```

### Dunder (magic) methods are special methods in Python with double underscores (__like_this__) that allow us to customize the behavior of built-in operations for our own objects.

They let your custom objects behave like built-in types.

Examples:

__add__ → lets you use + with your objects

__str__ → lets print() display your object nicely

__len__ → lets len() work on your object

__getitem__ → lets you use indexing (obj[0])


Operation (print, +, len, []) → Dunder method (__str__, __add__, __len__, __getitem__) → Your object behavior

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(2, 3)
print(p.x, p.y)  # Output: 2 3
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Point({self.x}, {self.y})"

p = Point(2, 3)
print(p)  # Output: Point(2, 3)
```

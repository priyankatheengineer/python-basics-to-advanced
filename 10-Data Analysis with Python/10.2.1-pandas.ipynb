```python
import pandas as pd
pd.__version__
```

```python
## Series ia pandas 1D labelled array that can hold data type.
data = [100,200,104]
series = pd.Series(data)
print(series)
```

```python
data = ["A", "B", "C"]
series = pd.Series(data, index=["a", "b", "c"])
print(series)
```

```python
## Accesing series elements using loc()

print(series.loc["a"])
print(series.loc["c"])
```

```python
## accessing series elements using iloc()

print(series.iloc[1])
```

```python
data = [100, 543, 234, 105, 213, 671]
series = pd.Series(data)
print(series)

print(series[series > 200])
```

```python
calories = {"Day 1": 1750, "Day 2": 2100, "Day 3": 1700}
series = pd.Series(calories)
print(series)
```

```python
print(series.loc["Day 1"])
```

```python
series.loc["Day 3"] += 500
print(series)
```

```python
## DataFrame - A tabular data structure with rows and columns. (2 Dimensional)

data = {"Name": ["Songebob", "Patrick", "Squidward"],
        "Age": [30, 35, 50]}

df = pd.DataFrame(data)
print(df)
```

```python
df = pd.DataFrame(data, index=["Emp 1", "Emp 2", "Emp 3"])
print(df)
```

```python
print(df.iloc[0])
```

```python
## add a new column
df["Job"] = ["Cook", "N/A", "Cashier"]

print(df)
```

```python
## add a new column
new_row = pd.DataFrame([{"Name": "Sandy", "Age": 45, "Job": "Engineer"}], index=["Emp 4"])
df = pd.concat([df, new_row])
print(df)
```

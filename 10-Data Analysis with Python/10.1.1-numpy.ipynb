```python
## Broadcasting - Broadcasting in NumPy is a mechanism that allows arrays of different shapes 
#                   to perform element-wise operations without explicitly reshaping them and 
#                   virtually expanding dimensions by matching the larger array's shape
# Same → OK
# One is 1 → OK
# Different and none is 1 → ERROR

# Broadcasting works when each dimension pair is either:
# equal
# or one of them is 1

# (2,3) &(2,3) - Valid → element-wise operation.
# (2,3) & (1,3) - broadcast result - (2,3)
# (2,3) & (3,) -> (2,3) & (1,3) broadcast result - (2,3)
# (1,4) & (4,1) broadcast result - (4,4)
# (3,) & (2,) (1,3) & (1,2) broadcast not possible

array1 = np.array([[1,2,3,4]])
array2 = np.array([[1],[2],[3],[4]])
print(array1.shape)
print(array2.shape)
print(array1)
print(array2)
print(array1+array2)

```

```python
ages = np.array([[21,17,19,20,16,30,18,65],
                 [39,22,15,99,18,19,20,21]])

teenagers = ages[ages < 18]
adults = ages[(ages >= 18) & (ages < 65)]
seniors = ages[ages >= 65]

evens = ages[ages % 2 == 0]
odds = ages[ages % 2 != 0]

print(teenagers)
print(adults)
print(seniors)
print(evens)
print(odds)
```

```python
## np.where(condition, value_if_true, value_if_false)

adults = np.where(ages >= 18, ages,0)
print(adults)
```
